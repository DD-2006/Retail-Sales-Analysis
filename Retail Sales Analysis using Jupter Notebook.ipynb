```python
orderss_df['order_date'] = pd.to_datetime(orderss_df['order_date'])
```

```python
reference_date = orderss_df['order_date'].max()
```

```python
merge_col=pd.merge(orderdetail_df,orderss_df,on="order_id",how='inner')
```

```python
rfm = merge_col.groupby('customer_id').agg(
    Recency=('order_date', 'max'),
    Frequency=('order_id', 'nunique'),
    Monetary=('total_price', 'sum')
)
rfm = rfm[['Recency', 'Frequency', 'Monetary']]

rfm['Recency'] = (reference_date - rfm['Recency']).dt.days
```

```python
rfm=rfm.reset_index()
```

```python
rfm.head()
```

```python
print(rfm[['Recency', 'Frequency', 'Monetary']].describe())
```

```python
rfm['Frequency'].value_counts()
```

```python
def segment_customer(row):
    if row['Recency']<=300 and row['Frequency']==3 and row['Monetary']>= 8000:
        return 'Champions'
    elif row['Recency']<=500 and row['Frequency']>=2 and row['Monetary']>= 5000 :
        return 'Loyal Customers'
    elif row['Recency']<=700 and row['Frequency']>=1 and row['Monetary']>= 3000:
        return 'Potential Loyalists'
    elif row['Recency']>700 and row['Monetary']<3000:
        return 'At Risk'
    
    else:
        return 'Others'

rfm['Segment'] = rfm.apply(segment_customer, axis=1)
```

```python
rfm
```

```python
rfm.to_sql(
    'customer_segments',
    con=engine,
    if_exists='replace',
    index=False
)
```
